### Q8: Time frame for a given country

Write a function that given a country name prints out the start and end date of their rankings.

Once you have created the function, call it for a few countries to test it.

```python
def show_timeframe(data,country):
    weeks = data[data['country_name']==country]["week"]
    start = weeks.min()
    end = weeks.max()
    print (f"Netflix rankings for {country} exist between {start} and {end}.")

show_timeframe(df,"Chile")
```

### Q9: Is number of weeks equal for each country?

The total number of weeks in the rankings is `len(df['week'].unique())`.

The total number of countries is `len(df['country_name'].unique())`.  

For every week, each country has 20 rows.   

If they all stayed the same number of weeks, then the product of these three weeks would be the same as the total number of rows.  

```python
df.groupby("country_name")["week"].nunique().nunique()==1
```

### Q11: New exercise

I first create a dicionary, then return the value (score) of the key (rank).

```python
dictionary = {1:25,2:18,3:15,4:12,5:10,6:8,7:6,8:4,9:2,10:1}
def rank2score (rank):
    return dictionary[rank]
```

Then, I add the score column using the above function.

```python
df["score"] = df["weekly_rank"].apply(rank2score)
```

```python
df.head()
```

Next, I will reproduce the table in task 2 description.

First, I will filter the show.

```python
lovely = df[df["show_title"] == "Lovely Runner"]
```

Then, I build the table using `agg()`.

```python
messy_table = lovely.groupby("country_name").agg(total_score = ("score","sum"),total_weeks=("week","nunique"),rankings_list=("weekly_rank",list))
```

I use `reset_index()` to turn country names into a regular column, then sort countries by their cumulative F1 scores in descending order.

```python
clean_table = messy_table.reset_index().sort_values("total_score", ascending=False)
clean_table
```

For my final project, our group plans to compare audience responses to wealthy CEO/heir male leads and male leads with ordinary occupations. I selected four K-dramas representing these two categories to explore their global popularity using Netflix Top 10 rankings.

First, I will check whether all four dramas appear in the dataset before comparing their performance across countries.

```python
dramas = ["Business Proposal", "King the Land", "Hometown Cha-Cha-Cha", "Welcome to Samdal-ri"]

df[df["show_title"].isin(dramas)]["show_title"].unique()
```

All four selected K-dramas are included in the Netflix Top 10 dataset. I will next compare their popularity across countries using the F1 scoring system, where higher rankings receive more points.

```python
business = df[df["show_title"] == "Business Proposal"]

business_messy_table = business.groupby("country_name").agg(
    total_score=("score", "sum"),
    total_weeks=("week", "nunique"),
    rankings_list=("weekly_rank", list)
)
business_clean_table = business_messy_table.reset_index().sort_values("total_score", ascending=False)
business_clean_table.head(10)
```

```python
king = df[df["show_title"] == "King the Land"]


king_messy_table = king.groupby("country_name").agg(
    total_score=("score", "sum"),
    total_weeks=("week", "nunique"),
    rankings_list=("weekly_rank", list)
)
king_clean_table = king_messy_table.reset_index().sort_values("total_score", ascending=False)
king_clean_table.head(10)

```

```python
hometown = df[df["show_title"] == "Hometown Cha-Cha-Cha"]

hometown_messy_table = hometown.groupby("country_name").agg(
    total_score=("score", "sum"),
    total_weeks=("week", "nunique"),
    rankings_list=("weekly_rank", list)
)
hometown_clean_table = hometown_messy_table.reset_index().sort_values("total_score", ascending=False)
hometown_clean_table.head(10)
```

```python
welcome = df[df["show_title"] == "Welcome to Samdal-ri"]

welcome_messy_table = welcome.groupby("country_name").agg(
    total_score=("score", "sum"),
    total_weeks=("week", "nunique"),
    rankings_list=("weekly_rank", list)
)
welcome_clean_table = welcome_messy_table.reset_index().sort_values("total_score", ascending=False)
welcome_clean_table.head(10)
```
